## Step 1: Verify Python Environment

```python
import sys
print(f"Python version: {sys.version}")
print(f"Python executable: {sys.executable}")
```

## Step 2: Check Required Libraries

```python
# Check if required libraries are installed
required_packages = ['pandas', 'numpy', 'matplotlib', 'seaborn', 'scikit-learn']

missing = []
installed = []

for package in required_packages:
    try:
        __import__(package)
        installed.append(package)
        print(f"✓ {package}")
    except ImportError:
        missing.append(package)
        print(f"✗ {package} - NOT INSTALLED")

if missing:
    print(f"\n⚠️ Missing packages: {', '.join(missing)}")
    print(f"Install with: pip install {' '.join(missing)}")
else:
    print("\n✅ All required packages are installed!")
```

## Step 3: Verify Data Files

```python
from pathlib import Path
import os

# Check data directory
data_dir = Path('../data/raw')

print(f"Data directory: {data_dir.absolute()}\n")

if data_dir.exists():
    files = list(data_dir.glob('*'))
    if files:
        print(f"Found {len(files)} files:\n")
        for file in sorted(files):
            if file.name != '.gitkeep':
                size_mb = file.stat().st_size / 1024 / 1024
                print(f"  ✓ {file.name:50s} {size_mb:6.1f} MB")
    else:
        print("⚠️ No data files found!")
else:
    print("❌ Data directory not found!")
```

## Step 4: Load Sample Data

```python
import pandas as pd

# Try loading the main CSV file
try:
    df = pd.read_csv('../data/raw/products_for_ts_grouped.csv')
    print(f"✅ Successfully loaded data!")
    print(f"\nShape: {df.shape[0]:,} rows × {df.shape[1]} columns")
    print(f"\nColumns: {list(df.columns)}")
    print(f"\nFirst few rows:")
    display(df.head())
except FileNotFoundError:
    print("⚠️ Main CSV file not found. Run notebook 01_explore_data.ipynb first to create it.")
except Exception as e:
    print(f"❌ Error loading data: {e}")
```

```python
if 'df' in locals():
    print("📊 Data Summary:\n")
    print(f"Date range: {df['date'].min()} to {df['date'].max()}")
    print(f"Unique products: {df['id_produit'].nunique():,}")
    print(f"Total records: {len(df):,}")
    
    if 'quantite_demande' in df.columns:
        print(f"\nDemand statistics:")
        print(f"  Min: {df['quantite_demande'].min():,.0f}")
        print(f"  Max: {df['quantite_demande'].max():,.0f}")
        print(f"  Mean: {df['quantite_demande'].mean():,.2f}")
        print(f"  Total: {df['quantite_demande'].sum():,.0f}")
```
