# Storage Optimization - Excel to CSV Converter

This notebook provides functionality to extract individual sheets from an Excel file and save them as CSV files.

## Usage:
1. Place your Excel (.xlsx) file in this folder
2. Run the cells to list available sheets
3. Use the `save_sheet_as_csv()` function to export any sheet to CSV

```python
import pandas as pd
import os
from pathlib import Path

# Get the current folder
current_folder = Path().absolute()
print(f"Working directory: {current_folder}")

# List all Excel files in the current folder
excel_files = list(current_folder.glob("*.xlsx"))
print(f"\nFound {len(excel_files)} Excel file(s):")
for file in excel_files:
    print(f"  - {file.name}")
```

```python
def save_sheet_as_csv(excel_file, sheet_name, output_csv=None):
    """
    Read a specific sheet from an Excel file and save it as CSV.
    
    Parameters:
    -----------
    excel_file : str
        Path to the Excel file (e.g., 'data.xlsx')
    sheet_name : str
        Name of the sheet to export
    output_csv : str, optional
        Output CSV filename. If None, uses '{sheet_name}.csv'
    
    Returns:
    --------
    str : Path to the saved CSV file
    """
    # Read the specific sheet
    df = pd.read_excel(excel_file, sheet_name=sheet_name)
    
    # Generate output filename if not provided
    if output_csv is None:
        output_csv = f"{sheet_name}.csv"
    
    # Save to CSV
    df.to_csv(output_csv, index=False)
    
    print(f"✓ Sheet '{sheet_name}' saved to '{output_csv}'")
    print(f"  Rows: {len(df)}, Columns: {len(df.columns)}")
    
    return output_csv
```

```python
# Example: List all sheets in an Excel file
# Replace 'your_file.xlsx' with your actual Excel filename

excel_filename = "WMS_Hackathon_DataPack_Templates_FR_FV_B7_ONLY.xlsx"  # Change this to your Excel file name

if os.path.exists(excel_filename):
    # Get all sheet names
    excel_file = pd.ExcelFile(excel_filename)
    sheet_names = excel_file.sheet_names
    
    print(f"Available sheets in '{excel_filename}':")
    for i, sheet in enumerate(sheet_names, 1):
        print(f"  {i}. {sheet}")
else:
    print(f"⚠ File '{excel_filename}' not found in current directory")
    print("Please update 'excel_filename' variable with the correct filename")
```
